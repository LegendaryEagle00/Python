# importar a biblioteca SQLite

```python
import sqlite3
```

## Cria uma conexão com o banco de dados (ou cria um novo se não existir)

```python
conn = sqlite3.connect('usuarios.db')
print("Banco de Dados Aberto com sucesso")
```

## Cria um cursor para interagir com o banco de dados

```python
cursor = conn.cursor()
print("Conectado ao Banco de dados")
```

##Criar Tabelas

```python
cursor.execute(
    '''
    CREATE TABLE IF NOT EXISTS usuarios(
        id INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT,
        nome TEXT NOT NULL,
        idade INTEGER NOT NULL,
        email TEXT NOT NULL,
        senha TEXT NOT NULL
    )
    '''
)
conn.commit()
```

##Inserir Dados

```python
# Insere dados na tabela "usuarios"
cursor.execute('''
INSERT INTO usuarios (nome, idade, email, senha) VALUES (?, ?, ?, ?)
''', ('Alice', 30, 'alice@teste.com', '12345678'))

cursor.execute('''
INSERT INTO usuarios (nome, idade, email, senha) VALUES (?, ?, ?, ?)
''', ('Bob', 25, 'bob@teste.com', '12345678'))

# Salva as mudanças
conn.commit()
```

##Consultar Dados

```python
# Consulta todos os usuários
cursor.execute('SELECT * FROM usuarios')
usuarios = cursor.fetchall()

for usuario in usuarios:
    print(usuario)
```

##Atualizar Dados

```python
# Atualiza a idade de Bob
cursor.execute('''
UPDATE usuarios SET idade = ? WHERE nome = ?
''', (26, 'Bob'))

# Salva as mudanças
conn.commit()
```

```python
# Consulta todos os usuários
cursor.execute('SELECT * FROM usuarios')
usuarios = cursor.fetchall()

for usuario in usuarios:
    print(usuario)
```

##Deletar Dados

```python
# Deleta o usuário Alice
cursor.execute('''
DELETE FROM usuarios WHERE nome = ?
''', ('Alice',))

# Salva as mudanças
conn.commit()
```

```python
# Consulta todos os usuários
cursor.execute('SELECT * FROM usuarios')
usuarios = cursor.fetchall()

for usuario in usuarios:
    print(usuario)
```

##Fechar a Conexão

```python
# Fecha a conexão
conn.close()
```

#Teste

```python
import sqlite3
conn = sqlite3.connect('cadastro.db')
cursor = conn.cursor()
print('Banco de dados Aberto e Conctado')
```

```python
#Criar tabela
cursor.execute('''
    CREATE TABLE IF NOT EXISTS usuarios(
        id INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT,
        nome TEXT NOT NULL,
        curso TEXT NOT NULL,
        idade INTEGER NOT NULL,
        cpf TEXT NOT NULL
)
''')
conn.commit()
```

```python
def cadastrar(nome, curso, idade, cpf):
    cursor.execute('''
    INSERT INTO usuarios (nome, curso, idade, cpf) VALUES (?, ?, ?, ?)
    ''', (nome, curso, idade, cpf))
    conn.commit()

def consultar(cpf):
    cursor.execute('SELECT * FROM usuarios WHERE cpf = ?', (cpf,))
    return cursor.fetchone()

def atualizar(cpf, novo_nome, novo_curso, nova_idade):
    cursor.execute('''
    UPDATE usuarios
    SET nome = ?, curso = ?, idade = ?
    WHERE cpf = ?
    ''', (novo_nome, novo_curso, nova_idade, cpf))
    conn.commit()

def deletar(cpf):
    cursor.execute('DELETE FROM usuarios WHERE cpf = ?', (cpf,))
    conn.commit()
```

```python
while True:
  print('''
    Voce Deseja?
    1 - Cadastrar
    2 - Consultar
    3 - Atualizar
    4 - Deletar
    5 - Sair
  ''')
  opcao = int(input('Digite a opção desejada: '))
  if opcao == 1:
    nome = input('Digite seu nome: ')
    curso = input('Digite seu curso: ')
    idade = int(input('Digite sua idade: '))
    cpf = input('Digite seu CPF: ')
    cadastrar(nome, curso, idade, cpf)
    print('Cadastro realizado com sucesso!')
  elif opcao == 2:
    pesquisa = input('Digite o cpf do usuario que deseja consultar: ')
    usuario = consultar(pesquisa)
    if usuario:
      print(usuario)
    else:
      print('Usuario não encontrado')
  elif opcao == 3:
    pesquisa = input('Digite o CPF do usuário que deseja atualizar: ')
    usuario = consultar(pesquisa)
    if usuario:
        novo_nome = input('Digite o novo nome: ')
        novo_curso = input('Digite o novo curso: ')
        nova_idade = int(input('Digite a nova idade: '))
        atualizar(pesquisa, novo_nome, novo_curso, nova_idade)
        print('Usuário atualizado com sucesso!')
    else:
        print('Usuário não encontrado.')
  elif opcao == 4:
    pesquisa = input('Digite o CPF do usuário que deseja deletar: ')
    usuario = consultar(pesquisa)
    if usuario:
        deletar(pesquisa)
        print('Usuário deletado com sucesso!')
    else:
        print('Usuário não encontrado.')
  elif opcao == 5:
    break
  else:
    print('Opção inválida. Tente novamente.')
```
