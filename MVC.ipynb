# MVC (Model-View-Controller)

O padrão MVC (Model-View-Controller) é uma arquitetura de software que separa uma aplicação em três componentes principais, facilitando a organização e manutenção do código.



##Model

 Representa a lógica de dados e as regras de negócio da aplicação. Ele gerencia a interação com a base de dados e é responsável por manipular os dados que a aplicação utiliza. O model notifica a view sobre mudanças, permitindo a atualização das informações exibidas.

```python
# model.py
class UserModel:
    def __init__(self):
        self.users = []

    def add_user(self, nome, curso, idade, cpf):
        user = {'nome': nome, 'curso': curso, 'idade': idade, 'cpf': cpf}
        self.users.append(user)

    def get_users(self):
        return self.users

    def find_user_by_cpf(self, cpf):
        for user in self.users:
            if user['cpf'] == cpf:
                return user
        return None
```

##View

É a interface do usuário, responsável pela apresentação dos dados. A view exibe as informações do model e envia as interações do usuário para o controller. Ela deve ser independente do model, permitindo que mudanças na lógica de dados não afetem a apresentação.

```python
# view.py
class UserView:
    @staticmethod
    def show_users(users):
        if users:
            for user in users:
                print(f"Nome: {user['nome']}, Curso: {user['curso']}, Idade: {user['idade']}, CPF: {user['cpf']}")
        else:
            print("Nenhum usuário cadastrado.")

    @staticmethod
    def user_not_found():
        print("Usuário não encontrado.")
```

##Controller

Atua como um intermediário entre o model e a view. O controller processa as entradas do usuário, chama as funções apropriadas no model e atualiza a view com os resultados. Ele recebe as ações do usuário (como cliques ou comandos) e as traduz em chamadas para o model.

```python
# controller.py
class UserController:
    def __init__(self, model, view):
        self.model = model
        self.view = view

    def add_user(self, nome, curso, idade, cpf):
        self.model.add_user(nome, curso, idade, cpf)

    def display_users(self):
        users = self.model.get_users()
        self.view.show_users(users)

    def find_user(self, cpf):
        user = self.model.find_user_by_cpf(cpf)
        if user:
            print(user)
        else:
            self.view.user_not_found()
```

##Usando o MVC

Ponto de entrada da aplicação, onde instanciamos o model, a view e o controller.

```python
# main.py

user_model = UserModel()
user_view = UserView()
user_controller = UserController(user_model, user_view)

# Adicionando usuários
user_controller.add_user("Alice", "Engenharia", 20, "12345678900")
user_controller.add_user("Bob", "Matemática", 22, "09876543211")

# Exibindo usuários
user_controller.display_users()

# Buscando um usuário
user_controller.find_user("12345678900")  # Usuário encontrado
user_controller.find_user("00000000000")   # Usuário não encontrado
```
