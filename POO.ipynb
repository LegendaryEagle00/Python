#Programação Orientada a Objetos (POO)

Programação Orientada a Objetos (POO) é um paradigma que permite organizar o código de forma mais modular e reutilizável, utilizando "objetos" que podem conter dados e funcionalidades.

##Classes e Objetos

- Classe: É um modelo ou molde que define as propriedades e comportamentos dos objetos.
- Objeto: É uma instância de uma classe, com suas próprias características.

```python
class Carro:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def dirigir(self):
        print(f"Dirigindo um {self.marca} {self.modelo}.")

# Criando um objeto da classe Carro
meu_carro = Carro("Toyota", "Corolla")
meu_carro.dirigir()
```

##Encapsulamento

O encapsulamento é o princípio de ocultar os detalhes internos de uma classe e expor apenas o que é necessário. Você pode usar modificadores de acesso, como:

- Público: Atributos e métodos acessíveis de fora da classe.
- Privado: Atributos e métodos que começam com __ e não são acessíveis diretamente fora da classe.

```python
class ContaBancaria:
    def __init__(self, saldo_inicial):
        self.__saldo = saldo_inicial  # Atributo privado

    def depositar(self, valor):
        self.__saldo += valor

    def consultar_saldo(self):
        return self.__saldo

conta = ContaBancaria(100)
conta.depositar(50)
print(conta.consultar_saldo())
```

## Herança

Herança permite que uma nova classe (subclasse) herde propriedades e métodos de uma classe existente (superclasse).

```python
class Veiculo:
    def __init__(self, tipo):
        self.tipo = tipo

    def mover(self):
        print(f"O {self.tipo} está se movendo.")

class Moto(Veiculo):
    def __init__(self, marca):
        super().__init__("moto")
        self.marca = marca

moto = Moto("Honda")
moto.mover()
```

##Polimorfismo

Polimorfismo permite que diferentes classes implementem métodos com o mesmo nome, mas com comportamentos diferentes.

```python
class Animal:
    def fazer_som(self):
        pass

class Cachorro(Animal):
    def fazer_som(self):
        return "Au Au"

class Gato(Animal):
    def fazer_som(self):
        return "Miau"

def emitir_som(animal):
    print(animal.fazer_som())

emitir_som(Cachorro())
emitir_som(Gato())
```

##Abstração

Abstração é o conceito de esconder a complexidade e mostrar apenas a interface necessária. Em Python, você pode usar classes abstratas do módulo `abc.`

```python
from abc import ABC, abstractmethod

class Forma(ABC):
    @abstractmethod
    def area(self):
        pass

class Retangulo(Forma):
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura

    def area(self):
        return self.largura * self.altura

retangulo = Retangulo(5, 10)
print(retangulo.area())  # Saída: 50
```

##Decoradores

Decoradores são uma maneira de modificar ou aprimorar o comportamento de funções ou métodos de forma concisa e reutilizável. Eles são funções que "decoram" outras funções, permitindo adicionar funcionalidades antes ou depois da execução da função original.

```python
def meu_decorador(func):
    def wrapper():
        print("Algo está acontecendo antes da função ser chamada.")
        func()
        print("Algo está acontecendo depois da função ser chamada.")
    return wrapper

@meu_decorador
def dizer_oi():
    print("Oi!")

dizer_oi()
```

##Decoradores com Argumentos

Se a função decorada aceita argumentos, você pode fazer o decorador aceitar também:

```python
def decorador_com_argumentos(func):
    def wrapper(*args, **kwargs):
        print("Antes da função ser chamada.")
        resultado = func(*args, **kwargs)
        print("Depois da função ser chamada.")
        return resultado
    return wrapper

@decorador_com_argumentos
def multiplicar(a, b):
    return a * b

print(multiplicar(2, 3))
```

##Uso Comum de Decoradores

- Logging: Para registrar informações sobre a execução de funções.
- Verificação de Acesso: Para restringir o acesso a determinadas funções.
- Memorização: Para armazenar resultados de funções pesadas e evitar cálculos repetidos.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Chamando {func.__name__} com {args} e {kwargs}")
        return func(*args, **kwargs)
    return wrapper

@logger
def somar(a, b):
    return a + b

somar(5, 3)
```

##Decoradores Integrados

Python possui decoradores integrados, como:

@staticmethod: Para definir métodos estáticos em classes.
@classmethod: Para definir métodos de classe que podem acessar a classe e seus atributos.
@property: Para definir propriedades que podem ser acessadas como atributos.

##Encadeamento de Decoradores

pode-se aplicar múltiplos decoradores a uma função:

```python
def decorador_1(func):
    def wrapper():
        print("Decorador 1")
        return func()
    return wrapper

def decorador_2(func):
    def wrapper():
        print("Decorador 2")
        return func()
    return wrapper

@decorador_1
@decorador_2
def saudacao():
    print("Olá!")

saudacao()
```

#Wrapper( )

O termo "wrapper" se refere a uma função ou um objeto que "envolve" outro, adicionando funcionalidades ou modificando seu comportamento sem alterar o código original. No contexto de decoradores, o "wrapper" é uma função que encapsula a função original.

Função Wrapper em Decoradores
- Encapsulamento: O wrapper permite que você adicione código antes e/ou depois da chamada da função original.
- Modificação: Você pode modificar os argumentos ou o resultado da função original.
- Manutenção da Assinatura: O wrapper geralmente aceita *args e **kwargs para manter a compatibilidade com a função original.
